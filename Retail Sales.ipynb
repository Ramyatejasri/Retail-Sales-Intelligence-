```python
import pandas as pd
import matplotlib.pyplot as plt
data = {
    "Product": ["Laptop", "Mobile", "Tablet", "Laptop", "Mobile", "Tablet"],
    "Sales": [50000, 30000, 20000, 45000, 35000, 25000],
    "Profit": [10000, 6000, 4000, 9000, 7000, 5000],
    "Region": ["East", "West", "North", "East", "West", "North"]
}
df = pd.DataFrame(data)
print(df)
```

```python

# 1. Sales Graph
df.groupby("Product")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Product")
plt.show()

# 2. Profit Graph
df.groupby("Product")["Profit"].sum().plot(kind="bar")
plt.title("Profit by Product")
plt.show()

# 3. Region Graph
df.groupby("Region")["Sales"].sum().plot(kind="pie", autopct="%1.1f%%")
plt.title("Sales by Region")
plt.show()
```
